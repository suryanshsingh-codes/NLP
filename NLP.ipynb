* 1. import all the required libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

* now fetch the csv file

```python
df=pd.read_csv("train.txt", sep=';', header=None, names=['text', 'emotion'])
'''
read the csv file
use comma separator
to avoid our first value to become header, we used 'none'
i have given header column names values manually
'''

df
```

* since machie learning can only deal with numbers so i have to replace the string data of emotion into numbers

```python
# check for null values
df.isnull().sum()
```

```python
# check the unique values of emotion
df['emotion'].unique()
```

```python
# now make a dictionary of all this emotions and number them
unique_emotions=df['emotion'].unique() # this will create an array and i will save this into a variable
emotions_numbers={}  # a dictionary created
i=0
for emo in unique_emotions:
    emotions_numbers[emo]=i
    i+=1

df['emotion']=df['emotion'].map(emotions_numbers)

df
```

* now i wil do text cleaning

```python
# lowercasing
df['text']=df['text'].apply(lambda x: x.lower())
df
```

```python
# remove punctuations
import string # to use that function
# create a function to remove
def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))

df['text']=df['text'].apply(remove_punc)
df
```

```python
# removing numbers

# create a function to remove
def remove_numbers(txt):
  new=""
  for i in txt:
    if not i.isdigit():
      new+=i
  return new

df['text']=df['text'].apply(remove_numbers)
df
```

```python
# remove emojis

'''
we will use ascii values
'''

def remove_emojis(txt):
  new=""
  for i in txt:
    if i.isascii():
      new =new+i
  return new
df['text']=df['text'].apply(remove_emojis)
df
```

* remove stop words; we will use a library using NLTK: natural language toolkit

```python
'''
data with messages is called corpus: very large data collection
each individual word is called token'''


import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
nltk.download('punkt') # used to tokenize
from nltk.tokenize import word_tokenize
```

```python
# first identify the stopwords
stopwords=set(stopwords.words('english'))
stopwords
```

```python
df.loc(1)['text']
```

```python
# convert the list of all words of text into an array and check if it exist in stop word and then remove if it does
def remove(txt):
  words=txt.split()
  cleaned=[]
  for i in words:
    if i not in stopwords:
      cleaned.append(i)

  return ' '.join(cleaned)

df['text']=df['text'].apply(remove)
df
```

* now we have to split the text into 20:80 format

```python
# for that we will do train test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.2, random_state=42)
```

NOW WE HAVE TO DO MODEL CREATION BUT BEFORE THAT WE HVAE TO **DO TEXT VECTORIZATION**

* our text is into => x train and x test-->> so i have to convert this into vectors, i will use both bag of words as well as tf-idf
* our emotion is into => y train and y test-->>so i have to convert this into vectors, i will use both bag of words as well as tf-idf


```python
# imort CountVectorizer for bag of words, TfidfVectorizer for tf-idf
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

* 1. *bag of words*

```python
# vectorizer for CountVectorizer(bag of words) to  create the vocabulary
bow_vectorizer=CountVectorizer()
# send the x train data to the do the text vectorization|
X_train_bow=bow_vectorizer.fit_transform(X_train)

# TRANSFORM test data using SAME vocabulary (NO fit!)
X_test_bow = bow_vectorizer.transform(X_test)  # ⚠️ Use .transform(), NOT .fit_transform();	ALWAYS on test/validation data (uses learned vocabulary)
```

```python
# now create the model for bag of words-> i am using here naive bais model
from sklearn.naive_bayes import MultinomialNB # this naive bayes model is used for text classification
# to check the accuracy
from sklearn.metrics import accuracy_score
# initialize the model and create it
nb_model = MultinomialNB()
# send the data to the model to train it
nb_model.fit(X_train_bow, y_train)
```

```python
# predict the testing data
pred_bow = nb_model.predict(X_test_bow)
# check the accuraacy score on testing y and predcitciton
print(accuracy_score(y_test, pred_bow))
print(pred_bow)
```

* 2. *tf-idf vectorizer*

```python
# ============================================
# 1. TF-IDF Vectorization
# ============================================

# Initialize TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer()

# FIT on training data (learn vocabulary and IDF values)
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)

# TRANSFORM test data using SAME vocabulary (NO fit!)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

# ============================================
# 2. Train Naive Bayes Model
# ============================================

# Create and train model
nb2_model = MultinomialNB()
nb2_model.fit(X_train_tfidf, y_train)
```

```python
# ============================================
# 3. Predict and Evaluate
# ============================================

# Predict on test data
pred_tfidf = nb2_model.predict(X_test_tfidf)

# Accuracy Score
print(f"\nAccuracy: {accuracy_score(y_test, pred_tfidf)}")
```

* 3. tf idf using logistci model also

```python
# ============================================
# LOGISTIC REGRESSION WITH EVALUATION
# ============================================

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Initialize model
logistic_model = LogisticRegression(max_iter=1000, random_state=42)

# Train model
logistic_model.fit(X_train_tfidf, y_train)
```

```python
# Predict on test data
log_pred = logistic_model.predict(X_test_tfidf)
```

```python
# Calculate and display accuracy
accuracy = accuracy_score(y_test, log_pred)
print(f"🎯 Logistic Regression Accuracy: {accuracy}")
```
